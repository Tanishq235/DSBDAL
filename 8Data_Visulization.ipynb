```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Titanic dataset from seaborn

df = sns.load_dataset('titanic')

# Display first 5 rows

df.head()
```

```python
# Dataset dimensions

df.shape
```

```python
# Display column names

df.columns
```

```python
# Dataset information

df.info()
```

```python
# Check missing values

df.isnull().sum()
```

```python
# Statistical summary

df.describe()
```

```python
# Count plot for survived passengers

sns.countplot(
    x='survived',
    data=df
)

plt.title("Survival Count")

plt.show()
```

```python
# Count plot for passenger class

sns.countplot(
    x='pclass',
    data=df
)

plt.title("Passenger Class Distribution")

plt.show()
```

```python
# Survival count based on gender

sns.countplot(
    x='survived',
    hue='sex',
    data=df
)

plt.title("Survival Based on Gender")

plt.show()
```

```python
# Age distribution

sns.histplot(
    df['age'],
    kde=True
)

plt.title("Age Distribution")

plt.show()
```

```python
# Fare distribution histogram

sns.histplot(
    df['fare'],
    bins=30,
    kde=True
)

plt.title("Fare Distribution")

plt.xlabel("Fare")

plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Boxplot for fare based on passenger class

sns.boxplot(
    x='pclass',
    y='fare',
    data=df
)

plt.title("Fare Distribution by Passenger Class")

plt.show()
```
