```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load Titanic dataset

df = sns.load_dataset('titanic')

# Display first 5 rows

df.head()
```

```python
# Dataset dimensions

df.shape
```

```python
# Display column names

df.columns
```

```python
# Dataset information

df.info()
```

```python
# Check missing values

df.isnull().sum()
```

```python
# Box plot for age distribution
# based on gender and survival

plt.figure(figsize=(10,6))

sns.boxplot(
    x='sex',
    y='age',
    hue='survived',
    data=df
)

plt.title(
    "Age Distribution based on Gender and Survival"
)

plt.xlabel("Gender")

plt.ylabel("Age")

plt.show()
```

```python
# Statistical summary of age

df['age'].describe()
```
